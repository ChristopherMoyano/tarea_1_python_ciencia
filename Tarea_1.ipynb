Importar librerias utilizadas

```python
import os
import sys
```

Imprimir listado de libros

```python
ruta = "Libros_txt_utf-8"
libros = os.listdir(ruta)
contador = 1
## bucle para recorrer los libros en la carpeta y enumerarlos
for i in libros:
    print(str(contador) + '.-' + i)
    contador+=1
```

Seleccionar libro

```python
## input para pedirle al usuario que elija un libro mediante un numero
libro_elegido = int(input("elige un libro (del 1 al 6):"))
nueva_ruta = ruta + '/'+libros[libro_elegido-1]
print(nueva_ruta)
```

Abrir libro

```python
Libro = open(nueva_ruta,"rt", encoding="UTF-8")
contenido_libro = Libro.read()
```

# Estadisticas

Contar cantidad total de palabras del libro

```python
palabras = contenido_libro.split()
palabras_totales = len(palabras) ## calcular el numero total de palabras en el libro con len()
print("El libro contiene", palabras_totales, "palabras")
```

Contar cantidad total de palabras no repetidas del libro

```python
palabras_no_repetidas = [] ## lista para almacenar las palabras no repetidas

## ciclo para recorrer la lista
for i in range(palabras_totales):
    if palabras[i] not in palabras_no_repetidas: ## si la palabra no esta en la lista, se cuenta
        palabras_no_repetidas.append(palabras[i])
no_repetidas = len(palabras_no_repetidas)
print("El libro contiene", no_repetidas, "palabras no repetidas")
```

Contar cantidad total de caracteres con espacio y sin espacio del libro

```python
caracteres_con_espacio = len(contenido_libro)-contenido_libro.count("\n") ## contar los caracteres con len, count y sin considerar los saltos de linea
print("El libro contiene", caracteres_con_espacio, "caracteres con espacio")
caracteres_sin_espacio = caracteres_con_espacio - contenido_libro.count(" ") ## contar los caracteres con len, count y sin considerar espacios
print("El libro contiene", caracteres_sin_espacio, "caracteres sin espacio")
```

Contar cantidad total de parrafos del libro

```python
parrafos = contenido_libro.split('\n') ## dividir el libro en parrafos con salto de linea como delimitador
parrafos_validos = [p for p in parrafos if p.strip() != ''] ## filtrar los parrafos vacios (solo espacios en blanco) y guardar parrafos validos
print("El libro contiene", len(parrafos_validos), "parrafos")
```

```python
filtro = '.?-"!¡¿*/\'=#$%(){[}],:' ## caracteres a eliminar
filtro_aplicado = str.maketrans("","",filtro) ## tabla de traduccion para eliminar los caracteres con maketrans
```

Buscar palabra completa con distincion de mayusculas y minusculas y contar sus instancias

```python
palabra_buscada = str(input("palabra a buscar: "))
contador_con_distincion = 0
for i in palabras: ## recorrer palabras de la lista (obtenida anteriormente en el conteo de palabras totales)
    i = i.translate(filtro_aplicado) ## eliminar caracteres no deseados de la palabra
    if i == palabra_buscada:
        contador_con_distincion+=1
print("La palabra: "+ palabra_buscada +" aparece "+str(contador_con_distincion)+" veces")
```

```python
contador_sin_distincion = 0
palabra_buscada_lower = palabra_buscada.lower() ## convertir la palabra buscada a letras minusculas para eliminar la distincion
for i in palabras:
    i = i.translate(filtro_aplicado)
    palabra = i.lower() ## convertir la palabra guardada anteriormente a letras minusculas para eliminar la distincion
    if palabra == palabra_buscada_lower: ## si ambas palabras coinciden se suma 1 al contador
        contador_sin_distincion+=1
print("La palabra: "+ palabra_buscada_lower +" aparece "+str(contador_sin_distincion)+" veces")
```

# Reemplazo

Remplazar una palabra completa con distincion de de mayusculas y minusculas

```python
palabra_reemplazar = input("Palabra a reemplazar: ")
reemplazo = input("Nuevo texto: ")

def reemplazar_con_distincion(texto, buscar, reemplazo):
    lineas = texto.split("\n")  
    for i in range(len(lineas)): ## recorrer las lineas del texto
        palabras = lineas[i].split(" ") ## dividir las lineas en palabras
        for j in range(len(palabras)): ## recorrer las palabras
            if palabras[j] == buscar: ## si la palabra coincide con la ingresada se remplaza
                palabras[j] = reemplazo
        lineas[i] = " ".join(palabras)  
    return "\n".join(lineas) 

contenido_libro_reemplazado = reemplazar_con_distincion(contenido_libro, palabra_reemplazar, reemplazo)
with open(nueva_ruta, "w", encoding="UTF-8") as f:
    f.write(contenido_libro_reemplazado)

Libro = open(nueva_ruta,"rt", encoding="UTF-8")
contenido_libro = Libro.read()

print("Reemplazo con distinción completado y guardado.")
```

Remplazar una palabra completa sin distincion de de mayusculas y minusculas

```python
palabra_reemplazar = input("Palabra a reemplazar: ")
reemplazo = input("Nuevo texto: ")

palabra_reemplazar_lower = palabra_reemplazar.lower()

def reemplazar_sin_distincion(texto, buscar, reemplazo):
    lineas = texto.split("\n")  
    for i in range(len(lineas)):
        palabras = lineas[i].split(" ")  
        for j in range(len(palabras)):
            if palabras[j].lower() == buscar:
                palabras[j] = reemplazo
        lineas[i] = " ".join(palabras)
    return "\n".join(lineas) 

contenido_libro_reemplazado = reemplazar_sin_distincion(contenido_libro, palabra_reemplazar_lower, reemplazo)
with open(nueva_ruta, "w", encoding="UTF-8") as f:
    f.write(contenido_libro_reemplazado)

Libro = open(nueva_ruta,"rt", encoding="UTF-8")
contenido_libro = Libro.read()

print("Reemplazo sin distinción completado y guardado.")
```
