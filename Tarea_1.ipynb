Imprimir listado de libros

Abrir libro

```python
Libro = open(nueva_ruta,"rt", encoding="UTF-8")
contenido_libro = Libro.read()
```

# Estadisticas

Contar cantidad total de palabras del libro

```python
palabras = contenido_libro.split()
palabras_totales = len(palabras) 
print(palabras_totales)
```

Contar cantidad total de palabras no repetidas del libro

```python
palabras_no_repetidas = []
for i in range(palabras_totales):
    if palabras[i] not in palabras_no_repetidas:
        palabras_no_repetidas.append(palabras[i])
no_repetidas = len(palabras_no_repetidas)
print(no_repetidas)
```

Contar cantidad total de caracteres con espacio y sin espacio del libro

```python
caracteres_con_espacio = len(contenido_libro)-contenido_libro.count("\n")
print(caracteres_con_espacio)
caracteres_sin_espacio = caracteres_con_espacio - contenido_libro.count(" ")
print(caracteres_sin_espacio)
```

Contar cantidad total de parrafos del libro

```python
parrafos = contenido_libro.split('\n')
parrafos_validos = [p for p in parrafos if p.strip() != '']
print(len(parrafos_validos))
```

Filtrar caracteres especiales del libro

```python
filtro = '.?-"!¡¿*/\'=#$%(){[}],:'
filtro_aplicado = str.maketrans("","",filtro)

    
```

Buscar palabra completa con distincion de mayusculas y minusculas y contar sus instancias

```python
palabra_buscada = str(input("palabra a buscar: "))
contador_con_distincion = 0
for i in palabras:
    i = i.translate(filtro_aplicado)
    if i == palabra_buscada:
        contador_con_distincion+=1
print("la palabra "+ palabra_buscada +" aparece : "+str(contador_con_distincion)+" veces")
```

```python
contador_sin_distincion = 0
palabra_buscada_lower = palabra_buscada.lower()
for i in palabras:
    i = i.translate(filtro_aplicado)
    palabra = i.lower()
    if palabra == palabra_buscada_lower:
        contador_sin_distincion+=1
print(contador_sin_distincion)
```

# Reemplazo

```python
palabra_reemplazar = input("Palabra a reemplazar: ")
reemplazo = input("Nuevo texto: ")

def reemplazar_con_distincion(texto, buscar, reemplazo):
    lineas = texto.split("\n")  
    for i in range(len(lineas)):
        palabras = lineas[i].split(" ")  
        for j in range(len(palabras)):
            if palabras[j] == buscar: 
                palabras[j] = reemplazo
        lineas[i] = " ".join(palabras)  
    return "\n".join(lineas) 

contenido_libro_reemplazado = reemplazar_con_distincion(contenido_libro, palabra_reemplazar, reemplazo)
with open(nueva_ruta, "w", encoding="UTF-8") as f:
    f.write(contenido_libro_reemplazado)

Libro = open(nueva_ruta,"rt", encoding="UTF-8")
contenido_libro = Libro.read()

print("Reemplazo con distinción completado y guardado.")
```
